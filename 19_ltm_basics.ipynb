```python
from langgraph.store.memory import InMemoryStore
```

```python
# Create a Store
store = InMemoryStore()
```

```python
# Creating a namespace 
namespace1 = ("user","u1")
namespace2 = ("user", "u2")
```

# Create/Insert Memories

```python
store.put(namespace1, "1", {"data": "User like Fish Tots."} )
store.put(namespace1, "2", {"data": "User prefers dark mode!"} )

store.put(namespace2, "1", {"data": "User like Pasta"} )
store.put(namespace2, "2", {"data": "User prefers Python programming language"} )
```

# Retrieve Memories

```python
store.get(namespace2,"2")
```

# Retrieve all Memories

```python
items = store.search(namespace1)

for i in items:
    print(i.value)
```
